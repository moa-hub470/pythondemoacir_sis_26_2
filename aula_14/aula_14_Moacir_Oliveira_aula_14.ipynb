```python
import requests
import pandas as pd

cidades = [
    "Araucaria",
    "Curitiba",
    "Londrina",
    "Maringa",
    "Ponta Grossa"
]

lista_dados = []

for cidade in cidades:

    parametros = {
        "key": api_key,
        "q": cidade,
        "lang": "pt"
    }

    try:
        resposta = requests.get(
            link_api,
            params=parametros,
            timeout=10
        )

        resposta.raise_for_status()

        lista_dados.append(resposta.json())

    except requests.exceptions.RequestException as erro:
        print(f"Erro ao consultar {cidade}: {erro}")
```

```python
df_clima = pd.json_normalize(lista_dados, sep="_")

display(
    df_clima[
        [
            "location_name",
            "location_localtime",
            "current_temp_c",
            "current_humidity"
        ]
    ]
)
```

```python
colunas = {
    "location_name": "Cidade",
    "location_localtime": "Data_Hora",
    "current_temp_c": "Temperatura_C",
    "current_humidity": "Umidade_Pct",
    "current_condition_text": "Condicao"
}

df_tratado = df_clima[
    list(colunas.keys())
].rename(columns=colunas).copy()

# Removendo duplicatas
df_tratado = df_tratado.drop_duplicates(
    subset=["Cidade"]
)

# Conversão de tipos
df_tratado["Data_Hora"] = pd.to_datetime(
    df_tratado["Data_Hora"],
    errors="coerce"
)

df_tratado["Temperatura_C"] = pd.to_numeric(
    df_tratado["Temperatura_C"],
    errors="coerce"
)

df_tratado["Umidade_Pct"] = pd.to_numeric(
    df_tratado["Umidade_Pct"],
    errors="coerce"
)
```

```python
resumo_clima = (
    df_tratado
    .groupby("Condicao")["Temperatura_C"]
    .mean()
    .reset_index()
)

print(resumo_clima)
```

```python
import requests
from google.colab import userdata

API_KEY = userdata.get("WEATHER_API_KEY")

link_api = "https://api.weatherapi.com/v1/current.json"

cidades = [
    "Araucaria",
    "Curitiba",
    "Londrina",
    "Maringa",
    "Ponta Grossa",
    "Sao Paulo",
    "Rio de Janeiro",
    "Florianopolis",
    "Porto Alegre",
    "Salvador"
]

lista_dados = []
```

```python
for cidade in cidades:

    parametros = {
        "key": API_KEY,
        "q": cidade,
        "lang": "pt"
    }

    try:
        resposta = requests.get(
            link_api,
            params=parametros,
            timeout=10
        )

        print(f"Cidade: {cidade}")
        print(f"Status HTTP: {resposta.status_code}")

        resposta.raise_for_status()

        dados = resposta.json()

        lista_dados.append(dados)

        print("Consulta realizada com sucesso!")

    except requests.exceptions.Timeout:
        print(f"Tempo limite excedido para {cidade}")

    except requests.exceptions.HTTPError:
        print(f"Erro HTTP: {resposta.status_code}")

    except requests.exceptions.RequestException as erro:
        print(f"Falha na conexão: {erro}")

    print("-" * 40)
```

```python
if lista_dados:

    print(lista_dados[0].keys())

    for registro in lista_dados:

        cidade = registro["location"]["name"]
        temperatura = registro["current"]["temp_c"]
        condicao = registro["current"]["condition"]["text"]

        print(f"Cidade: {cidade}")
        print(f"Temperatura: {temperatura} °C")
        print(f"Condição: {condicao}")
        print("-" * 40)

else:
    print("Nenhum dado foi coletado.")
```

```python
import pandas as pd

df_clima = pd.json_normalize(lista_dados, sep="_")

colunas = {
    "location_name": "Cidade",
    "location_region": "Estado",
    "location_localtime": "Data_Hora",
    "current_temp_c": "Temperatura_C",
    "current_humidity": "Umidade_Pct",
    "current_condition_text": "Condicao"
}

df_tratado = df_clima[
    list(colunas.keys())
].rename(columns=colunas).copy()

display(df_tratado)
```

```python
df_tratado = df_tratado.drop_duplicates(
    subset=["Cidade"]
).copy()

df_tratado["Data_Hora"] = pd.to_datetime(
    df_tratado["Data_Hora"],
    errors="coerce"
)

df_tratado["Temperatura_C"] = pd.to_numeric(
    df_tratado["Temperatura_C"],
    errors="coerce"
)

df_tratado["Umidade_Pct"] = pd.to_numeric(
    df_tratado["Umidade_Pct"],
    errors="coerce"
)

df_tratado = df_tratado.dropna(
    subset=["Cidade", "Temperatura_C"]
)

df_tratado.info()

display(df_tratado.describe())

display(df_tratado)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.bar(
    df_tratado["Cidade"],
    df_tratado["Temperatura_C"],
    color="#E91E63"
)

plt.title(
    "Temperatura Atual nas Cidades Brasileiras",
    fontsize=14
)

plt.xlabel("Cidades", fontsize=12)
plt.ylabel("Temperatura (°C)", fontsize=12)

for i, temperatura in enumerate(df_tratado["Temperatura_C"]):
    plt.text(
        i,
        temperatura + 0.3,
        f"{temperatura:.1f}°C",
        ha="center"
    )

plt.xticks(rotation=45)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.5
)

plt.tight_layout()
plt.show()
```

```python
df_tratado.to_csv(
    "clima_cidades_brasileiras.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Arquivo CSV exportado com sucesso!")
```
